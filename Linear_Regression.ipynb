```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
url = "https://raw.githubusercontent.com/ybifoundation/Dataset/main/Salary%20Data.csv?utm_source=chatgpt.com"

df = pd.read_csv(url)
```

```python
df.columns
```

```python
df.rename(columns={'Experience Years':'Year'},inplace=True)
```

```python
df.columns
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
plt.scatter(df['Year'],df['Salary'])
```

```python
X = df.iloc[:,0:1]
Y = df.iloc[:,1]
```

```python
Y.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,Y,test_size=0.2)
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
y_pred
```

```python
X_test
```

```python
print("R2 Score: ", r2_score(y_test,y_pred))
```

```python
y_test
```

```python
y_all_pred = lr.predict(X)
```

```python
plt.scatter(X,Y)
plt.plot(X,y_all_pred,color="red")
```

```python
m = lr.coef_
m
```

```python
c = lr.intercept_
c
```

```python
x=int(input("Enter the Year_Experience:"))
salary = m*x+c
print("The Salary is predicted",salary)
```
